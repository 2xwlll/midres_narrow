```python
from pathlib import Path
from blimpy import Waterfall


def inspect_file(path, label):
    print("\n" + "="*70)
    print(label)
    print(path)
    print("="*70)

    wf = Waterfall(str(path), load_data=False)

    h = wf.header

    print("\nHeader:")
    for key in [
        "fch1",
        "foff",
        "nchans",
        "tsamp",
        "tstart",
        "nbits",
    ]:
        print(f"{key:10s}: {h.get(key)}")

    print("\nExpected data dimensions:")
    print(
        "time samples:",
        h.get("n_ints_in_file")
    )
    print(
        "channels:",
        h.get("nchans")
    )


# -------------------------
# Voyager
# -------------------------

voyager = Path(
    "/datag/public/voyager_2020/"
    "spliced_blc00010203040506o7o0111213141516o7o021222324252627_guppi_59046_81310_DIAG_VOYAGER-1_0015.rawspec.0002.h5"
)

inspect_file(
    voyager,
    "VOYAGER MIDRES"
)


# -------------------------
# CARMENES OFF
# -------------------------

offlist = Path(
    "/datax/scratch/wlll2x/"
    "carmen_midres/cadence_000/off.lst"
)

with open(offlist) as f:
    carmen_file = next(
        line.strip()
        for line in f
        if line.strip()
    )


inspect_file(
    Path(carmen_file),
    "CARMENES OFF MIDRES"
)
```
